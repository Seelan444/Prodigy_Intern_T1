```python
import pandas as pd
df = pd.read_csv("population_by_age_group.csv")
print(df.head())
print(df.columns)
```

```python
import altair as alt
# Ensure 'Country' column is of type string
df['Country'] = df['Country'].astype(str)

# Create the Altair plot (Bar Chart)
chart = alt.Chart(df).mark_bar().encode(
    x=alt.X('Country:N', title='Country'),  # Nominal data type for Country
    y=alt.Y('Total:Q', title='Population'),  # Quantitative data type for Total
    tooltip=['Country', 'Total']  # Add tooltips for interactivity
).properties(
    title='Population by Country (Bar Chart)'
)

# Save chart as PNG
chart.save('altair_chart.png')

# Display the image in the notebook
from IPython.display import Image
Image(filename='altair_chart.png')
```
